## Primer conjunto de datos

Un DataFrame es una tabla que podemos explorar y transformar con pandas.

```python
ventas = pd.DataFrame({
    "producto": ["Café", "Té", "Taza", "Termo"],
    "unidades": [12, 18, 7, 5],
    "precio": [85, 60, 120, 250]
})

ventas
```

```python
ventas["importe"] = ventas["unidades"] * ventas["precio"]
ventas.sort_values("importe", ascending=False)
```
