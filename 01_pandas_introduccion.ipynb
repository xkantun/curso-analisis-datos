# Introducción a pandas: explorar ventas

En esta actividad cargaremos datos desde un archivo CSV y construiremos un pequeño análisis exploratorio.

```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
ruta = Path("../datos/ventas.csv")
ventas = pd.read_csv(ruta, parse_dates=["fecha"])
ventas.head()
```

## 1. Inspección inicial

```python
print("Filas y columnas:", ventas.shape)
ventas.info()
ventas.describe(numeric_only=True)
```

## 2. Variables derivadas

Calcularemos el importe total de cada registro y el mes de la venta.

```python
ventas["importe"] = ventas["unidades"] * ventas["precio_unitario"]
ventas["mes"] = ventas["fecha"].dt.to_period("M").astype(str)

resumen_producto = (
    ventas.groupby("producto", as_index=False)["importe"]
    .sum()
    .sort_values("importe", ascending=False)
)

resumen_producto
```

```python
plt.figure(figsize=(8, 4))
sns.barplot(data=resumen_producto, x="producto", y="importe", hue="producto", legend=False)
plt.title("Importe acumulado por producto")
plt.ylabel("Importe")
plt.xlabel("")
plt.show()
```
